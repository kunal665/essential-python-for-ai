# Session 2: Operators and Conditional Statements

A hands-on look at Python's operators — arithmetic, comparison, logical, membership, identity, and assignment — followed by how to make decisions in code with `if`/`elif`/`else`.

## 1. A Bit More on Variable Assignment

Before we get to operators, two assignment shortcuts you'll see constantly in real code.

**Multiple assignment** — assign several related variables in one line. Hyperparameters are often defined together this way:

````python
learning_rate, batch_size = 0.001, 32
print(learning_rate, batch_size)
````

**Chained assignment** — assign the same starting value to multiple variables at once. A common case: initializing two counters to zero before you start tallying results.

````python
correct_predictions = total_predictions = 0
print(correct_predictions, total_predictions)
````

### Variable naming rules

- Names must start with a letter or underscore (`_`), never a digit.
- Digits are allowed after the first character: `first_score` is valid, `1st_score` is not.
- By convention, constants (values that shouldn't change) are written in `ALL_CAPS`, e.g. `MAX_EPOCHS`.
- Python's style guide is **PEP 8** — it recommends `snake_case` (`batch_size`) over `camelCase` (`batchSize`).

## 2. The Boolean Type

You've already seen `int`, `float`, `str`, `list`, `tuple`, `dict`, and `set`. There's one more core type you'll use constantly once you start writing comparisons and conditions: **`bool`**.

A `bool` has exactly two values: `True` and `False`. Every comparison you write (`5 > 3`, `accuracy >= 0.9`) evaluates to one of these.

````python
has_model_converged = False
print(has_model_converged, type(has_model_converged))
````

## 3. Operators: An Overview

Python groups operators into six categories:

| Category | Operators | What they do |
|---|---|---|
| Arithmetic | `+ - * / // % **` | Perform math |
| Comparison | `== != < > <= >=` | Compare two values, return a `bool` |
| Logical | `and or not` | Combine `bool` values |
| Membership | `in`, `not in` | Check if a value exists inside a collection |
| Identity | `is`, `is not` | Check if two variables point to the *same object* |
| Assignment | `= += -= *= /=` | Assign or update a variable's value |

## 4. Arithmetic Operators

Most of these are familiar from basic math. Two are Python-specific:
- `//` is **floor division** — divides, then rounds down.
- `%` is **modulo** — gives you the remainder.

These two show up together constantly when splitting data into batches — you need to know both how many *full* batches you get, and how many samples are left over.

````python
total_samples = 100
batch_size = 32

print(total_samples / batch_size)    # 3.125 -- exact division
print(total_samples // batch_size)   # 3     -- number of full batches you can form
print(total_samples % batch_size)    # 4     -- samples left over, not enough for one more batch
````

`**` is the power operator:

````python
prediction_error = 4
squared_error = prediction_error ** 2
print(squared_error)
````

## 5. Comparison Operators

Comparison operators compare two values and always return a `bool`.

````python
model_a_accuracy = 0.85
model_b_accuracy = 0.90

print(model_a_accuracy == model_b_accuracy)
print(model_a_accuracy != model_b_accuracy)
print(model_a_accuracy < model_b_accuracy)
````

Comparison also works on lists — Python checks them element by element:

````python
predicted_labels = [1, 0, 1]
actual_labels = [1, 0, 0]

print(predicted_labels == actual_labels)
````

Note that this only tells you whether the two lists are *exactly* identical — it's not how you'd actually measure a model's accuracy (that means counting how many positions match, which needs a loop — a later topic). It's just showing you how `==` behaves on a list.

## 6. Logical Operators

`and`, `or`, and `not` combine `bool` values so you can express more than one condition at once.

- **`and`** — `True` only if both sides are `True`.
- **`or`** — `True` if at least one side is `True`.
- **`not`** — flips a `bool`.

A realistic use — deciding whether a model is ready to ship, which usually depends on more than one thing being true at once:

````python
model_accuracy = 0.92
inference_time_ms = 150

is_accurate_enough = model_accuracy >= 0.90
is_fast_enough = inference_time_ms <= 200

print(is_accurate_enough and is_fast_enough)   # ship only if both hold
print(is_accurate_enough or is_fast_enough)    # true if at least one holds
print(not is_accurate_enough)
````

## 7. Membership Operators

`in` and `not in` check whether a value exists inside a collection. To use them, the collection has to be **iterable** — something Python can step through element by element. Roughly: `str`, `list`, `tuple`, `dict`, and `set` are iterable; `int`, `float`, and `bool` are not.

````python
valid_classes = ["cat", "dog", "bird"]

print("dog" in valid_classes)
print("fish" not in valid_classes)
````

Here's what happens on a non-iterable value:

````python
confidence_score = 0.87
print(0.5 in confidence_score)   # raises TypeError -- a float isn't iterable
````

**A dictionary gotcha:** `in` on a `dict` checks its **keys**, not its values. This matters the moment you start working with config dictionaries.

````python
model_config = {"learning_rate": 0.01, "batch_size": 32, "epochs": 10}

print(0.01 in model_config)                # False -- 0.01 is a value, not a key
print("batch_size" in model_config)        # True -- this is a key
print(0.01 in model_config.values())       # True -- this checks values instead
````

🎯 **AI/ML relevance:** Code that checks `if "learning_rate" in config:` is everywhere in ML pipelines. If you actually meant to check whether a *value* exists (e.g. "is this label one of the valid outputs"), forgetting this distinction gives you a silent `False` instead of catching a real bug.

## 8. Identity Operators

`is` and `is not` check whether two variables point to the **exact same object in memory** — not whether they hold equal values. This is different from `==`, which only checks value.

Recall from Session 1: small integers are cached, so equal small numbers always share an `id()`. Larger numbers aren't cached.

````python
run_one_final_epoch = 1000
run_two_final_epoch = 1000

print(run_one_final_epoch == run_two_final_epoch)   # True -- same value
print(run_one_final_epoch is run_two_final_epoch)   # False -- separate objects, not cached
````

The same distinction applies to mutable types like lists:

````python
experiment_one_batch_sizes = [16, 32, 64]
experiment_two_batch_sizes = [16, 32, 64]

print(experiment_one_batch_sizes == experiment_two_batch_sizes)   # True -- same values
print(experiment_one_batch_sizes is experiment_two_batch_sizes)   # False -- two different objects
````

Because they're genuinely separate objects, modifying one doesn't touch the other:

````python
experiment_one_batch_sizes.append(128)
print(experiment_one_batch_sizes)   # [16, 32, 64, 128]
print(experiment_two_batch_sizes)   # unchanged -- [16, 32, 64]
````

Now the version that actually causes bugs — assigning with `=` doesn't create a copy, it gives a second name to the *same* object:

````python
default_hyperparameters = {"learning_rate": 0.01, "batch_size": 32}

model_one_config = default_hyperparameters   # NOT a copy -- same dict
model_two_config = default_hyperparameters   # also the same dict

print(model_one_config is model_two_config)   # True

model_one_config["learning_rate"] = 0.1
print(model_two_config["learning_rate"])      # 0.1 -- changed too, even though we never touched model_two_config directly
````

🎯 **AI/ML relevance:** This is a real bug pattern — sharing a config or hyperparameter dict across two "separate" model instances by assignment, then being confused when tuning one affects the other. `is` is also the correct way to check for `None`: always write `if value is None:`, not `if value == None:`.

## 9. Assignment Operators

Beyond plain `=`, Python has shorthand operators that update a variable using its current value — this is exactly how you'd accumulate a running total, e.g. counting predictions as they come in:

````python
correct_predictions = total_predictions = 0

total_predictions += 1     # a new sample just came in
print(total_predictions)

correct_predictions += 1   # and the model got it right
print(correct_predictions)
````

The same shorthand works for `-=`, `*=`, `/=`. One note if you're coming from C, Java, or JavaScript: Python has **no `++` or `--` operators** — always use `+= 1` / `-= 1`.

## 10. Conditional Statements: `if` / `elif` / `else`

Everything so far produces a value. Conditional statements let your program **make decisions** — run different code depending on whether something is `True` or `False`.

Python uses **indentation**, not `{ }`, to mark blocks of code. Everything indented under an `if` (by convention, 4 spaces — the PEP 8 standard) belongs to that block, and a block can contain as many statements as you need.

The general shape:
```
if condition:
    # runs if condition is True
elif another_condition:
    # runs if the first was False and this is True
else:
    # runs if none of the above were True
```
`elif` and `else` are both optional.

````python
model_a_accuracy = 0.87
model_b_accuracy = 0.91

if model_a_accuracy > model_b_accuracy:
    print("Model A performs better")
elif model_b_accuracy > model_a_accuracy:
    print("Model B performs better")
    print("Recommending Model B for deployment")
else:
    print("Both models perform equally")
````

You could check membership in a list of valid classes with a chain of `or`s (`predicted_class == "cat" or predicted_class == "dog" or ...`), but the membership operator you already know does this more cleanly and scales to any number of classes without adding more `or`s:

````python
predicted_label = "fish"
valid_classes = ["cat", "dog", "bird"]

if predicted_label in valid_classes:
    print("Valid class")
else:
    print("Invalid class")
````

A more realistic branching example — checking whether a model's overall accuracy clears a bar, after many predictions have already been tallied. Note the threshold is a named variable rather than a hardcoded number buried in the condition — that keeps the code self-documenting and easy to change later.

````python
correct_predictions = 350
total_predictions = 500
accuracy_percentage = (correct_predictions / total_predictions) * 100
minimum_accuracy_threshold = 70

if accuracy_percentage > minimum_accuracy_threshold:
    print("Model meets accuracy bar")
else:
    print("Model needs more training")
````

## 11. Truthy and Falsy Values

An `if` doesn't strictly require a `bool` — it accepts *any* value, and Python decides whether to treat it as `True` or `False`.

**Falsy:** `False`, `0`, `0.0`, `[]`, `''`, `()`, `{}`, `None`
**Truthy:** everything else, including any non-empty collection or non-zero number.

````python
model_predictions = [0.2, 0.8, 0.1]

if model_predictions:
    print("Truthy")
else:
    print("Falsy")
````

````python
model_predictions = []

if model_predictions:
    print("Truthy")
else:
    print("Falsy")
````

🎯 **AI/ML relevance:** This is exactly why you'll see `if not predictions:` in real code instead of `if len(predictions) == 0:`. It's the idiomatic way to check whether a list, batch, or query result came back empty — and it's also how a function that should have returned `None` on failure can silently slip past a check that only tests truthiness, if you're not aware of this rule.
