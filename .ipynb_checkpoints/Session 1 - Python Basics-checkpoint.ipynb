## 2. Variables and Assignment

A variable is just a name that points to a value stored in memory. You create one using `=`.

````python
language = "Python"
learner_age = 25
print(language, learner_age)
````

You can reassign a variable to a new value at any time — even a value of a different type:

````python
favorite_number = 10
print(favorite_number)

favorite_number = "now I'm text"   # reassigned to a different type — legal, but usually a sign to rename
print(favorite_number)
````

## 3. Data Types

Every value in Python has a type. Here are the six core ones you'll use constantly:

| Type | Example | Description |
|---|---|---|
| `int` | `10` | whole numbers |
| `float` | `15.88` | decimal numbers |
| `str` | `"Python"` | text |
| `list` | `[70, 80, 90]` | an ordered collection you can change later |
| `tuple` | `(100, 200, 300)` | an ordered collection meant to stay fixed |
| `dict` | `{"name": "Asha"}` | labeled data, as `key: value` pairs |
| `set` | `{10, 20, 30}` | a collection with no duplicates, no fixed order |

````python
total_students = 10                          # int
apple_price = 15.88                          # float
course_name = "Python"                       # str
exam_scores = [70, 80, 90, 88, 100]          # list
image_shape = (100, 200, 300)                # tuple
student_profile = {"name": "Asha", "email": "asha@example.com"}   # dict
unique_scores = {10, 20, 30, 40, 50}         # set
````

Use `type()` to check the type of any value — useful whenever you're not sure what you're working with:

````python
print(exam_scores, type(exam_scores))
print(student_profile, type(student_profile))
````

**Quick sense of when you'd reach for which:**
- `list` — an ordered collection you might change later (add/remove items).
- `tuple` — an ordered collection that should *stay fixed*.
- `dict` — data with labels (`key: value` pairs), like a record.
- `set` — a collection with no duplicates and no particular order.

## 4. Mutable vs Immutable (the concept, in plain terms)

Some values can be changed "in place" after they're created — Python doesn't need to create a new object, it just updates the existing one in memory. Others cannot — any "change" actually creates a brand-new object elsewhere in memory.

- **Mutable** (can change in place): `list`, `dict`, `set`
- **Immutable** (cannot change in place): `int`, `float`, `str`, `tuple`

This matters practically: if you hand a mutable value to a function and that function changes it, the change is visible everywhere else that value is used too — because it's still the *same* object underneath. This is one of the most common sources of confusing bugs for beginners, so it's worth understanding *why* it happens, not just memorizing the rule.

## 5. Proving it: Object Identity with `id()`

`id()` shows you the memory address of an object. If two variables give the same `id()`, they're pointing to the exact same object in memory.

**Immutable — reassignment creates a new object:**

````python
course_name = "Python"
print(id(course_name))

course_name = "Java"          # this does NOT change the string "Python" — it points course_name at a NEW object
print(id(course_name))
````

**Mutable — modifying in place keeps the same object:**

````python
exam_scores = [70, 80, 90, 88]
print(id(exam_scores))

exam_scores.append(95)        # modifies the SAME list object
print(id(exam_scores))
print(exam_scores)
````

**A small quirk of Python you'll notice while testing this:** small integers (from -5 to 256) are cached internally, so two variables holding the same small number will always show the same `id()`. This isn't something you need to rely on in real code — it's just useful for understanding what `id()` is actually showing you.

````python
attendance_count = 100
attendance_count_copy = 100
print(id(attendance_count), id(attendance_count_copy))   # same -- small int, cached

registered_users = 1000
registered_users_copy = 1000
print(id(registered_users), id(registered_users_copy))   # different -- large int, not cached
````

## 6. Putting It Together: Why Mutability Matters in Practice

````python
def add_item(shopping_cart):
    shopping_cart.append("bread")
    return shopping_cart

grocery_list = ["milk", "eggs"]
add_item(grocery_list)
print(grocery_list)   # changed even though we never reassigned grocery_list ourselves
````
